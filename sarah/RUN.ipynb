```python
from marketsim.wrappers.multi_asset_wrapper import MultiAssetEnv
from marketsim.data.load_historical import load_two_asset_series

csv_path = r"R:\sescott1\Masters\thesis\potential_codes\StockMARL\Stock-MARL-main\resources\datasets\train_dataV1.csv"

historical_series = load_two_asset_series(
    csv_path=csv_path,
    ticker_a="AAPL",
    ticker_b="XOM",
    sim_time=50,
)

env = MultiAssetEnv(
    historical_series=historical_series,
    sim_time=50,
    max_decision_events=10,
    num_background_agents=5,
    lam_bg=0.10,
    lam_rl=0.05,
    q_max=10,
    pv_var=1.0,
    zi_shade=[0.05, 0.5],
    initial_cash=100_000.0,
    lambda_invalid=1.0,
    use_market_makers=True,
    lam_mm=0.10,
    mm_xi=0.5,
    mm_K=3,
    mm_omega=2.0,
    seed=42,
)

obs, info = env.reset(seed=42)
print("obs shape:", obs.shape)
print("initial obs:", obs)

for _ in range(5):
    a = env.action_space.sample()
    obs, reward, terminated, truncated, info = env.step(a)
    print("a:", a, "reward:", reward, "done:", terminated or truncated, "info:", info)
    if terminated or truncated:
        break
```

```python
for ticker in env.tickers:
    market = env.markets[ticker]
    print(
        ticker,
        "bid=", market.order_book.get_best_bid(),
        "ask=", market.order_book.get_best_ask()
    )
```
